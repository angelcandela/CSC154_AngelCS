```python
#Import the BeautifulSoup library so we can parse HTML and XML documents.
```

```python
import urllib.request
from bs4 import BeautifulSoup
url = "https://steamcharts.com/top"
headers = {'User-Agent': 'Mozilla/5.0'}
req = urllib.request.Request(url, headers=headers)
page = urllib.request.urlopen(req)
soup = BeautifulSoup(page, "html.parser")
print(soup.prettify())
```

```python
#Use the find_all() method to get all instances of the table tag
```

```python
all_tables = soup.find_all("table")
print(len(all_tables))
for table in all_tables:
    print("TABLE: ", table, "\n")
```

```python
#this line grabs that entire <table> block so you can loop through the rows later.
```

```python
right_table = soup.find("table", class_="common-table")
```

```python
#append data to list .
```

```python
rank = []
game = []
current_players = []
peak_players = []
hours_played = []

for row in right_table.find_all('tr')[1:]:  
    cells = row.find_all('td')
    if len(cells) == 6:
        rank.append(cells[0].text.strip())
        game.append(cells[1].find('a').text.strip())
        current_players.append(cells[2].text.strip().replace(',', ''))
        peak_players.append(cells[4].text.strip().replace(',', ''))
        hours_played.append(cells[5].text.strip().replace(',', ''))
```

```python
#organize data with their correct spot.
```

```python
import pandas as pd

df = pd.DataFrame({
    'Rank': rank,
    'Game': game,
    'Current Players': current_players,
    'Peak Players': peak_players,
    'Hours Played': hours_played
})

print(df)
```
